```python
import importlib
import subprocess
import sys

try:
    plt = importlib.import_module("matplotlib.pyplot")
    nx = importlib.import_module("networkx")
except ModuleNotFoundError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "matplotlib", "networkx"])
    plt = importlib.import_module("matplotlib.pyplot")
    nx = importlib.import_module("networkx")

def crear_nodo(valor):
    return {
        "valor": valor,
        "izquierda": None,
        "derecha": None,
        "altura": 1
    }

def insertar_bst(nodoact, valornue):
    if nodoact is None:
        return crear_nodo(valornue)

    if valornue < nodoact["valor"]:
        nodoact["izquierda"] = insertar_bst(nodoact["izquierda"], valornue)
    elif valornue > nodoact["valor"]:
        nodoact["derecha"] = insertar_bst(nodoact["derecha"], valornue)

    act_alt(nodoact)
    return nodoact

def altura(nodoact):
    if nodoact is None:
        return 0
    return nodoact["altura"]

def act_alt(nodoact):
    altizq = altura(nodoact["izquierda"])
    altder = altura(nodoact["derecha"])
    nodoact["altura"] = 1 + max(altizq, altder)

def fac_bal(nodoact):
    if nodoact is None:
        return 0
    return altura(nodoact["izquierda"]) - altura(nodoact["derecha"])

def rot_der(nodoact):
    nuevopad = nodoact["izquierda"]
    subcent = nuevopad["derecha"]

    nuevopad["derecha"] = nodoact
    nodoact["izquierda"] = subcent

    act_alt(nodoact)
    act_alt(nuevopad)

    return nuevopad

def rot_izq(nodoact):
    nuevopad = nodoact["derecha"]
    subcent = nuevopad["izquierda"]

    nuevopad["izquierda"] = nodoact
    nodoact["derecha"] = subcent

    act_alt(nodoact)
    act_alt(nuevopad)

    return nuevopad

def insertar_avl(nodoact, valornue):
    if nodoact is None:
        return crear_nodo(valornue)

    if valornue < nodoact["valor"]:
        nodoact["izquierda"] = insertar_avl(nodoact["izquierda"], valornue)
    elif valornue > nodoact["valor"]:
        nodoact["derecha"] = insertar_avl(nodoact["derecha"], valornue)
    else:
        return nodoact

    act_alt(nodoact)
    balance = fac_bal(nodoact)

    if balance > 1 and valornue < nodoact["izquierda"]["valor"]:
        return rot_der(nodoact)

    if balance < -1 and valornue > nodoact["derecha"]["valor"]:
        return rot_izq(nodoact)

    if balance > 1 and valornue > nodoact["izquierda"]["valor"]:
        nodoact["izquierda"] = rot_izq(nodoact["izquierda"])
        return rot_der(nodoact)

    if balance < -1 and valornue < nodoact["derecha"]["valor"]:
        nodoact["derecha"] = rot_der(nodoact["derecha"])
        return rot_izq(nodoact)

    return nodoact

def preorden(nodoact, recorrido):
    if nodoact is not None:
        recorrido.append(nodoact["valor"])
        preorden(nodoact["izquierda"], recorrido)
        preorden(nodoact["derecha"], recorrido)

def inorden(nodoact, recorrido):
    if nodoact is not None:
        inorden(nodoact["izquierda"], recorrido)
        recorrido.append(nodoact["valor"])
        inorden(nodoact["derecha"], recorrido)

def postorden(nodoact, recorrido):
    if nodoact is not None:
        postorden(nodoact["izquierda"], recorrido)
        postorden(nodoact["derecha"], recorrido)
        recorrido.append(nodoact["valor"])

def agregar_nodos(nodoact, graf, posiciones, lados, posx, posy, separacion):
    if nodoact is None:
        return

    valoract = nodoact["valor"]
    posiciones[valoract] = (posx, posy)
    graf.add_node(valoract)

    nodoizq = nodoact["izquierda"]
    nododer = nodoact["derecha"]

    if nodoizq is not None:
        valorizq = nodoizq["valor"]
        graf.add_edge(valoract, valorizq)
        lados[(valoract, valorizq)] = "Izq"
        agregar_nodos(nodoizq, graf, posiciones, lados, posx - separacion, posy - 1, separacion / 2)

    if nododer is not None:
        valorder = nododer["valor"]
        graf.add_edge(valoract, valorder)
        lados[(valoract, valorder)] = "Der"
        agregar_nodos(nododer, graf, posiciones, lados, posx + separacion, posy - 1, separacion / 2)

def preparar_grafo(raiz):
    graf = nx.DiGraph()
    posiciones = {}
    lados = {}
    agregar_nodos(raiz, graf, posiciones, lados, 0, 0, 4)
    return graf, posiciones, lados

def dibujar_grafo(raiz, titulo, eje):
    graf, posiciones, lados = preparar_grafo(raiz)
    coloresnod = []
    coloresari = []

    for valornodo in graf.nodes():
        if valornodo == raiz["valor"]:
            coloresnod.append("gold")
        else:
            coloresnod.append("skyblue")

    for arista in graf.edges():
        if lados[arista] == "Izq":
            coloresari.append("forestgreen")
        else:
            coloresari.append("darkorange")

    nx.draw_networkx_nodes(graf, posiciones, node_color=coloresnod, node_size=1400, ax=eje)
    nx.draw_networkx_labels(graf, posiciones, font_size=10, ax=eje)
    nx.draw_networkx_edges(graf, posiciones, edge_color=coloresari, arrows=True, arrowsize=18, ax=eje)
    nx.draw_networkx_edge_labels(graf, posiciones, edge_labels=lados, font_color="black", ax=eje)
    eje.set_title(titulo)
    eje.axis("off")

def mostrar_arbol(raiz, titulo):
    figura, eje = plt.subplots(figsize=(8, 5))
    dibujar_grafo(raiz, titulo, eje)
    figura.tight_layout()
    plt.show()

def comparar_arboles(raizbst, raizavl):
    figura, ejes = plt.subplots(1, 2, figsize=(11, 5))
    dibujar_grafo(raizbst, "BST", ejes[0])
    dibujar_grafo(raizavl, "AVL", ejes[1])
    figura.suptitle("Comparacion entre BST y AVL", fontsize=14, y=0.98)
    figura.tight_layout(rect=[0, 0, 1, 0.92])
    plt.show()
    print("Altura BST:", altura(raizbst))
    print("Altura AVL:", altura(raizavl))

cantprod = 0

while cantprod <= 0:
    try:
        cantprod = int(input("Cantidad de productos: "))
        if cantprod <= 0:
            print("La cantidad debe ser mayor que cero.")
    except ValueError:
        print("Ingrese un numero entero valido.")

codigos = []

for numprod in range(1, cantprod + 1):
    codval = False

    while not codval:
        try:
            codprod = int(input("Ingrese el codigo del producto " + str(numprod) + ": "))
            codigos.append(codprod)
            codval = True
        except ValueError:
            print("Ingrese un codigo numerico valido.")

raizbst = None
raizavl = None

for codprod in codigos:
    raizbst = insertar_bst(raizbst, codprod)
    raizavl = insertar_avl(raizavl, codprod)

print("\nCodigos ingresados:", codigos)

mostrar_arbol(raizbst, "Arbol BST")
mostrar_arbol(raizavl, "Arbol AVL")
comparar_arboles(raizbst, raizavl)

prebst = []
preavl = []
preorden(raizbst, prebst)
preorden(raizavl, preavl)

inbst = []
inavl = []
inorden(raizbst, inbst)
inorden(raizavl, inavl)

postbst = []
postavl = []
postorden(raizbst, postbst)
postorden(raizavl, postavl)

print("\nPreorden BST:", prebst)
print("Preorden AVL:", preavl)
print("\nInorden BST:", inbst)
print("Inorden AVL:", inavl)
print("\nPostorden BST:", postbst)
print("Postorden AVL:", postavl)
```
