### 5) Análisis de complejidad **[15 pts]**
Incluye:
- Justificación
- Función T[n]
- Mejor y peor caso con gráfica

Sin memoización, la recursión simple (T(n) = T(n-1) + T(n-2)) recalcula los mismos subproblemas una y otra vez por lo que el número de llamadas crece de forma exponencial aproximadamente O(2ⁿ) Con memoización, cada valor de T(k) para k entre 0 y n se calcula una sola vez y se guarda las siguientes veces que se necesita se consulta en O(1) en vez de recalcularlo. Esto reduce el algoritmo a O(n) en tiempo.

Función T[n]

T(n) = 0, si n = 0
T(n) = 1, si n = 1
T(n) = T(n-1) + T(n-2), si n ≥ 2

Con memoización, el costo computacional C(n) (número de operaciones) sigue:

C(n) = C(n-1) + O(1), ya que T(n-2) ya está en memoria cuando se calcula T(n-1)
Esto se resuelve como C(n) = O(n)



Mejor y peor caso

Mejor caso: n = 0 o n = 1 entonces O(1), porque son casos base y se resuelven directamente sin recursión.
Peor caso: n grande entonces O(n) en tiempo y O(n) en espacio por la tabla de memoria y la pila de llamadas recursivas


## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
# Escribe tu códigodef calcular_plantas(n, memo=None):
def calcular_plantas(n, memo=None):
    
    if memo is None:
        memo = {}

   
    if n in memo:
        return memo[n]

   
    if n == 0:
        return 0
    if n == 1:
        return 1

    
    resultado = calcular_plantas(n - 1, memo) + calcular_plantas(n - 2, memo)

    
    memo[n] = resultado

    
    return resultado


def pedir_mes():
    
    while True:
        entrada = input("Ingrese el mes (n, debe ser un número entero mayor o igual a 0): ")
        if entrada.isdigit():
            return int(entrada)
        print("Valor inválido. Por favor ingrese un número entero mayor o igual a 0.")



n = pedir_mes()
resultado = calcular_plantas(n)


print(f"Mes consultado: {n}")
print(f"Nuevas plantas producidas: {resultado}")
```
